```python
dbutils.widgets.removeAll()
```

```python
%sql
create widget text storageName default "adbsmartdata0311prod";
create widget text catalogwk default "catalog_tr";
```

```python
%python
storageName = dbutils.widgets.get("storageName")
```

```python
%sql
CREATE CATALOG IF NOT EXISTS $catalogwk
COMMENT 'Catalogo para la arquitectura medallion del ambiente de prod';
```

```python
%sql
DROP SCHEMA IF EXISTS $catalogwk.raw;
DROP SCHEMA IF EXISTS $catalogwk.bronze;
DROP SCHEMA IF EXISTS $catalogwk.silver;
DROP SCHEMA IF EXISTS $catalogwk.golden;
```

```python
%python
dbutils.fs.rm(f"abfss://bronze@{storageName}.dfs.core.windows.net/",True)
dbutils.fs.rm(f"abfss://silver@{storageName}.dfs.core.windows.net/",True)
dbutils.fs.rm(f"abfss://golden@{storageName}.dfs.core.windows.net/",True)
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS $catalogwk.raw;
CREATE SCHEMA IF NOT EXISTS $catalogwk.bronze;
CREATE SCHEMA IF NOT EXISTS $catalogwk.silver;
CREATE SCHEMA IF NOT EXISTS $catalogwk.golden;
```

###Tablas Bronze

```python
%sql
CREATE TABLE IF NOT EXISTS catalog_tr.bronze.transactions (
transaction_id string,
transaction_date timestamp,
customer_id string,
product_id string,
product_name string,
category string,
quantity integer,
unit_price double,
total_amount double,
store_location string,
payment_method string
)
USING DELTA
LOCATION "abfss://bronze@${storageName}.dfs.core.windows.net/transactions"
```
